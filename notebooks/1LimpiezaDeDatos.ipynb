```python
import pandas as pd
```

```python
df = pd.read_csv('/content/V-Dem-CY-Full+Others-v16.csv')
print(df.shape)
```

```python
# Identificación
id_cols = ["country_name", "country_id", "year"]

# Índices agregados de exclusión política
exclusion_cols = [
    "v2xpe_exlsocgr", "v2xpe_exlgender", "v2xpe_exlecon",
    "v2xpe_exlpol", "v2xpe_exlgeo"
]

# Distribución de poder por grupo
power_cols = [
    "v2pepwrsoc", "v2pepwrgen", "v2pepwrort", "v2pepwrses", "v2pepwrgeo"
]

# Igualdad ante la ley / acceso a justicia
legal_cols = [
    "v2clacjstm", "v2clacjstw", "v2cldmovem", "v2cldmovew",
    "v2clsocgrp", "v2x_clpriv"
]

# Represión de sociedad civil por identidad
repression_cols = [
    "v2csgender", "v2csantimv", "v2csrlgrep", "v2csrlgcon"
]

# Equidad en servicios básicos
equity_cols = [
    "v2peedueq", "v2pehealth", "v2peapsecon", "v2peapssoc"
]

# Género y sufragio
gender_suffrage_cols = [
    "v2x_gender", "v2x_suffr", "v2fsuffrage", "v2msuffrage"
]

# Variables de control/contexto
control_cols = [
    "v2x_polyarchy", "v2x_libdem", "e_gdppc", "e_regiongeo",
    "e_pop", "e_area"
]

# Desviación estándar de los índices clave (ángulo crítico: desacuerdo
# entre codificadores como proxy de fiabilidad del dato)
sd_cols = [
    "v2xpe_exlsocgr_sd", "v2xpe_exlgender_sd",
    "v2clsocgrp_sd", "v2x_gender_sd"
]

target_cols = (
    id_cols + exclusion_cols + power_cols + legal_cols +
    repression_cols + equity_cols + gender_suffrage_cols +
    control_cols + sd_cols
)

print(f"Total de columnas solicitadas: {len(target_cols)}")
```

```python
available = [c for c in target_cols if c in df.columns]
missing = [c for c in target_cols if c not in df.columns]

print(f"Encontradas: {len(available)}/{len(target_cols)}")
if missing:
    print("\nNO encontradas en este archivo:")
    for m in missing:
        print(f"  - {m}")
```

```python
df_subset = df[available].copy()
df_subset.head()
```

```python
missing_pct = (df_subset.isna().mean() * 100).sort_values(ascending=False)
print(missing_pct)

df_clean = df_subset[df_subset["year"] >= 2000].copy()
df_clean["year"] = df_clean["year"].astype(int)

print(f"Filas antes: {len(df_subset)} -> después del filtro de año: {len(df_clean)}")
```

```python
output_name = "vdem_exclusion_subset.csv"
df_clean.to_csv(output_name, index=False)
```
